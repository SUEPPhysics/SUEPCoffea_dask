```python
import h5py
import pickle
import numpy as np
import sys
sys.path.append('../../')
sys.path.append('../../plotting/')
import matplotlib.pyplot as plt
from plotting.utils.plot_utils import plot_ratio
from plotting.utils import plot_utils
from plotting.utils import hist_utils
import mplhep as hep

with open('kaons.pkl', 'rb') as f:
    data = pickle.load(f)
hists_kaons = data['hists']
hists = {}
for key in hists_kaons:
    hists[key] = hists_kaons[key].copy()

with open('new_leptonic.pkl', 'rb') as f:
    data = pickle.load(f)
hists_lep = data['hists']
for key in hists_lep:
    hists[key] += hists_lep[key]

#with open('new_generic.pkl', 'rb') as f:
with open('generic_noGenPions.pkl', 'rb') as f:
    data = pickle.load(f)
hists_gen = data['hists']
for key in hists_gen:
    hists[key] += hists_gen[key]
```

```python
pt_pi_m = hists_lep['pt_pi_matched']
pt_pi_u = hists_lep['pt_pi_unmatched']
pt_pi_m_gen = hists_gen['pt_pi_matched']
pt_pi_u_gen = hists_gen['pt_pi_unmatched']

new_bins = [1,2,3,4,5,7,10,12,15,30]
pt_pi_m = hist_utils.rebin_piecewise(pt_pi_m, new_bins)
pt_pi_u = hist_utils.rebin_piecewise(pt_pi_u, new_bins)
pt_pi_m_gen = hist_utils.rebin_piecewise(pt_pi_m_gen, new_bins)
pt_pi_u_gen = hist_utils.rebin_piecewise(pt_pi_u_gen, new_bins)

pi_eff_by_pt = pt_pi_m.values() / (pt_pi_m.values() + pt_pi_u.values())
err_pi_eff_by_pt = (pi_eff_by_pt * (1 - pi_eff_by_pt) / (pt_pi_m.values() + pt_pi_u.values()))**0.5
pi_eff_by_pt_gen = pt_pi_m_gen.values() / (pt_pi_m_gen.values() + pt_pi_u_gen.values())
err_pi_eff_by_pt_gen = (pi_eff_by_pt_gen * (1 - pi_eff_by_pt_gen) / (pt_pi_m_gen.values() + pt_pi_u_gen.values()))**0.5

fig = plt.figure(figsize=(11, 8))
ax = fig.add_subplot(111)
ax.scatter(pt_pi_m.axes[0].centers, pi_eff_by_pt, label="$\pi$ - $m_{A'} = 0.5 GeV$", marker='.')
ax.errorbar(pt_pi_u.axes[0].centers, pi_eff_by_pt, yerr=err_pi_eff_by_pt, fmt='.')
ax.scatter(pt_pi_m_gen.axes[0].centers, pi_eff_by_pt_gen, label="$\pi$ - $m_{A'} = 1.0 GeV$", marker='.')
ax.errorbar(pt_pi_u_gen.axes[0].centers, pi_eff_by_pt_gen, yerr=err_pi_eff_by_pt_gen, fmt='.')
ax.axhline(1.0, color='darkgray', linestyle='--')
ax.legend(loc=(1.01, 0))
ax.set_ylim(0, 1.05)
ax.set_xlabel('$p^{\mathrm{gen}}_T$ [GeV]')
ax.set_ylabel('Track Reconstruction Efficiency')
hep.cms.label(llabel='Preliminary', ax=ax)
fig.tight_layout()
```

```python
stats = {
    'e': {'matched': 0, 'unmatched': 0},
    'mu': {'matched': 0, 'unmatched': 0},
    'pi': {'matched': 0, 'unmatched': 0},
    'kaon': {'matched': 0, 'unmatched': 0},
}

particle_types = ['e', 'mu', 'pi', 'kaon']
for particle in particle_types:
    stats[particle]['matched'] = hists[f'pt_{particle}_matched'].sum().value
    stats[particle]['unmatched'] = hists[f'pt_{particle}_unmatched'].sum().value

    if stats[particle]['matched'] + stats[particle]['unmatched'] == 0:
        continue

    eff = stats[particle]['matched'] / (stats[particle]['matched'] + stats[particle]['unmatched'])
    err_eff = (eff * (1 - eff) / (stats[particle]['matched'] + stats[particle]['unmatched']))**0.5

    stats[particle]['eff'] = eff

    print(f'Efficiency for {particle}: {eff:.3f} +/- {err_eff:.3f}')
    print(f'Inefficiency for {particle}: {1 - eff:.3f}')

print("Total particles:")
for particle in particle_types:
    total = stats[particle]['matched'] + stats[particle]['unmatched']
    print(f"Total {particle}: {total}")

ineff_mu_over_pi = (1 - stats['mu']['eff']) / (1 - stats['pi']['eff'])
print("ineff mu / ineff pi:", round(ineff_mu_over_pi, 3))

ineff_e_over_pi = (1 - stats['e']['eff']) / (1 - stats['pi']['eff'])
print("ineff e / ineff pi:", round(ineff_e_over_pi, 3))

ineff_kaon_over_pi = (1 - stats['kaon']['eff']) / (1 - stats['pi']['eff'])
print("ineff K / ineff pi:", round(ineff_kaon_over_pi, 3))

generic_factor = 0.03*1.0
hadronic_factor = 0.03*0.85 + ineff_e_over_pi*0.03*0.15
leptonics_factor = 0.03*0.6 + ineff_e_over_pi*0.03*0.40

print("Tracking systematic values:")
print(f"Generic\t: {generic_factor:.3f}")
print(f"Hadronic: {hadronic_factor:.3f}")
print(f"Leptonic: {leptonics_factor:.3f}")
```

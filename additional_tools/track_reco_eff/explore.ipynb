```python
stats = {
    'e': {'matched': 0, 'unmatched': 0},
    'mu': {'matched': 0, 'unmatched': 0},
    'pi': {'matched': 0, 'unmatched': 0},
}
stats['e']['matched'] = hists['pt_e_matched'].sum().value
stats['e']['unmatched'] = hists['pt_e_unmatched'].sum().value
stats['mu']['matched'] = hists['pt_mu_matched'].sum().value
stats['mu']['unmatched'] = hists['pt_mu_unmatched'].sum().value
stats['pi']['matched'] = hists['pt_pi_matched'].sum().value
stats['pi']['unmatched'] = hists['pt_pi_unmatched'].sum().value

eff_e = stats['e']['matched'] / (stats['e']['matched'] + stats['e']['unmatched'])
err_eff_e = (eff_e * (1 - eff_e) / (stats['e']['matched'] + stats['e']['unmatched']))**0.5

eff_mu = stats['mu']['matched'] / (stats['mu']['matched'] + stats['mu']['unmatched'])
err_eff_mu = (eff_mu * (1 - eff_mu) / (stats['mu']['matched'] + stats['mu']['unmatched']))**0.5

eff_pi = stats['pi']['matched'] / (stats['pi']['matched'] + stats['pi']['unmatched'])
err_eff_pi = (eff_pi * (1 - eff_pi) / (stats['pi']['matched'] + stats['pi']['unmatched']))**0.5

print(f'Efficiency for electrons: {eff_e:.3f} +/- {err_eff_e:.3f}')
print(f'Efficiency for muons: {eff_mu:.3f} +/- {err_eff_mu:.3f}')
print(f'Efficiency for pions: {eff_pi:.3f} +/- {err_eff_pi:.3f}')

print(f'Inefficiency for electrons: {1 - eff_e:.3f}')
print(f'Inefficiency for muons: {1 - eff_mu:.3f}')
print(f'Inefficiency for pions: {1 - eff_pi:.3f}')

print("Total electrons:", stats['e']['matched'] + stats['e']['unmatched'])
print("Total muons:", stats['mu']['matched'] + stats['mu']['unmatched'])
print("Total pions:", stats['pi']['matched'] + stats['pi']['unmatched'])

ineff_e_over_pi = (1 - eff_e) / (1 - eff_pi)
print("ineff e / ineff pi:", ineff_e_over_pi)

generic_factor = 0.03*1.0
hadronic_factor = 0.03*0.85 + ineff_e_over_pi*0.03*0.15
leptonics_factor = 0.03*0.6 + ineff_e_over_pi*0.03*0.40

print("Tracking systematic values:")
print(f"Generic\t: {generic_factor:.3f}")
print(f"Hadronic: {hadronic_factor:.3f}")
print(f"Leptonic: {leptonics_factor:.3f}")
```
